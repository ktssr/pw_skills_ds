## 1. What is a Convolutional Neural Network (CNN), and how does it differ from traditional fully connected neural networks in terms of architecture and performance on image data?

- Convolutional Neural Network (CNN) is a specialized type of neural network designed for processing grid-like data such as images. CNNs are particularly effective for computer vision tasks. Following are the key concepts in CNNs:
  - Convolutional Layer: Uses filters/kernels to detect features like edges, textures, and patterns by sliding across the input image
  - Pooling Layer: Reduces spatial dimensions while retaining important information, helping to reduce computational cost
  - Feature Maps: Output of convolutional operations that highlight specific features in the input
  - Local Connectivity: Each neuron connects only to a small region of the input, unlike fully connected networks
  - Parameter Sharing: Same filter weights are used across different positions, reducing the number of parameters
  - Translation Invariance: Can detect features regardless of their position in the image

- Differences from traditional fully connected neural networks:
  - Architecture: CNNs use convolutional and pooling layers whereas fully connected networks use only dense layers
  - Parameters: CNNs have fewer parameters due to weight sharing, making them more efficient
  - Spatial Information: CNNs preserve spatial relationships in images while fully connected networks flatten the input
  - Performance: CNNs achieve better accuracy on image data due to their ability to capture local patterns and hierarchical features
  - Computational Efficiency: CNNs are more efficient for image processing due to local connectivity and parameter sharing

## 2. Discuss the architecture of LeNet-5 and explain how it laid the foundation for modern deep learning models in computer vision. Include references to its original research paper.

- LeNet-5 is a pioneering CNN architecture developed by Yann LeCun et al. in 1998. It was designed for handwritten digit recognition and laid the foundation for modern deep learning. Following are the key components of LeNet-5:
  - Input Layer: 32x32 grayscale images
  - C1 (Convolutional Layer): 6 feature maps of size 28x28 using 5x5 filters
  - S2 (Subsampling/Pooling): 6 feature maps of size 14x14 using 2x2 average pooling
  - C3 (Convolutional Layer): 16 feature maps of size 10x10 using 5x5 filters
  - S4 (Subsampling/Pooling): 16 feature maps of size 5x5 using 2x2 average pooling
  - C5 (Convolutional Layer): 120 feature maps of size 1x1 using 5x5 filters
  - F6 (Fully Connected): 84 neurons
  - Output Layer: 10 neurons for digit classification

- Foundation for modern deep learning:
  - Hierarchical Feature Learning: Demonstrated that CNNs can learn features automatically from low-level to high-level
  - Alternating Pattern: Established the pattern of alternating convolutional and pooling layers
  - End-to-End Learning: Showed that entire network can be trained using backpropagation
  - Practical Application: Proved CNNs work effectively on real-world problems like digit recognition

- Original Research Paper: "Gradient-Based Learning Applied to Document Recognition" by Y. LeCun, L. Bottou, Y. Bengio, and P. Haffner, published in Proceedings of the IEEE, 1998. This paper introduced LeNet-5 and demonstrated its effectiveness on the MNIST dataset.

## 3. Compare and contrast AlexNet and VGGNet in terms of design principles, number of parameters, and performance. Highlight key innovations and limitations of each.

- AlexNet (2012) and VGGNet (2014) are both landmark CNN architectures that significantly advanced computer vision. Following are the comparisons:

### Design Principles:
- AlexNet:
  - Used larger filters (11x11, 5x5) in early layers
  - Employed ReLU activation function
  - Used dropout for regularization
  - Applied data augmentation techniques
  - Utilized GPU acceleration for training

- VGGNet:
  - Used only small 3x3 filters throughout the network
  - Emphasized depth over filter size
  - Maintained consistent architecture pattern
  - Used multiple 3x3 filters to achieve larger receptive fields

### Number of Parameters:
- AlexNet: Approximately 60 million parameters
- VGGNet: VGG-16 has 138 million parameters, VGG-19 has 144 million parameters

### Performance:
- AlexNet: Achieved 15.3% top-5 error on ImageNet (2012)
- VGGNet: VGG-16 achieved 7.3% top-5 error on ImageNet (2014)

### Key Innovations:
- AlexNet:
  - First to use ReLU activation in CNNs
  - Introduced dropout for regularization
  - Demonstrated effectiveness of data augmentation
  - Showed importance of GPU acceleration

- VGGNet:
  - Proved that depth matters more than filter size
  - Simplified architecture with uniform 3x3 filters
  - Showed that stacking small filters is more effective than using large filters

### Limitations:
- AlexNet:
  - Large number of parameters in fully connected layers
  - Computationally expensive due to large filters
  - Limited depth compared to later architectures

- VGGNet:
  - Very large number of parameters leading to overfitting
  - High computational cost and memory requirements
  - Slow training and inference times

## 4. What is transfer learning in the context of image classification? Explain how it helps in reducing computational costs and improving model performance with limited data.

- Transfer learning is a machine learning technique where a model trained on one task is adapted for a related task. In image classification, it involves using a pre-trained CNN model and adapting it for a new classification problem. Following are the key concepts:
  - Pre-trained Models: Models trained on large datasets like ImageNet with millions of images
  - Feature Extraction: Using pre-trained layers as fixed feature extractors
  - Fine-tuning: Updating some or all layers of the pre-trained model on new data
  - Domain Adaptation: Adapting knowledge from source domain to target domain

- How it reduces computational costs:
  - Training Time: No need to train the entire network from scratch, significantly reducing training time
  - Hardware Requirements: Requires less computational power as most layers are already trained
  - Data Collection: Reduces the need for collecting massive amounts of labeled data
  - Energy Consumption: Lower energy consumption due to reduced training time

- How it improves performance with limited data:
  - Feature Reuse: Pre-trained models have already learned useful low-level features like edges and textures
  - Better Initialization: Starting with pre-trained weights provides better initialization than random weights
  - Regularization Effect: Pre-trained features act as regularizers, preventing overfitting on small datasets
  - Faster Convergence: Model converges faster as it starts from a good initial state
  - Higher Accuracy: Often achieves better performance than training from scratch, especially with limited data

## 5. Describe the role of residual connections in ResNet architecture. How do they address the vanishing gradient problem in deep CNNs?

- Residual connections (skip connections) are the key innovation in ResNet architecture that allows training of very deep neural networks. Following are the key concepts:
  - Skip Connections: Direct connections that bypass one or more layers
  - Residual Block: Basic building block containing skip connections
  - Identity Mapping: Skip connections perform identity mapping of the input
  - Residual Learning: Instead of learning H(x), the network learns F(x) = H(x) - x

- How residual connections work:
  - Input Addition: The input is added to the output of the convolutional layers
  - Shortcut Path: Provides an alternative path for gradient flow during backpropagation
  - Identity Function: If optimal function is identity, the network can easily learn to set weights to zero
  - Easier Optimization: Makes it easier to optimize the network by providing direct gradient paths

- Addressing vanishing gradient problem:
  - Gradient Flow: Skip connections provide direct paths for gradients to flow backward
  - Gradient Preservation: Gradients can flow through skip connections without being diminished
  - Deep Network Training: Enables training of networks with hundreds of layers
  - Stable Learning: Provides more stable gradient flow compared to traditional deep networks
  - Performance Improvement: Deeper networks with residual connections perform better than shallower ones

- Benefits of residual connections:
  - Deeper Networks: Allows training of much deeper networks (50, 101, 152 layers)
  - Better Performance: Achieves better accuracy with increased depth
  - Faster Training: Networks train faster due to better gradient flow
  - Reduced Degradation: Solves the degradation problem where deeper networks perform worse



## 6. Implement the LeNet-5 architectures using Tensorflow or PyTorch to classify the MNIST dataset. Report the accuracy and training time.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import time
import matplotlib.pyplot as plt

# Load and preprocess MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize pixel values to [0, 1]
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# Reshape to add channel dimension and resize to 32x32 (LeNet-5 input size)
x_train = tf.image.resize(x_train[..., tf.newaxis], [32, 32])
x_test = tf.image.resize(x_test[..., tf.newaxis], [32, 32])

# Convert labels to categorical
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)

print(f"Training data shape: {x_train.shape}")
print(f"Test data shape: {x_test.shape}")
```

```python
# Implement LeNet-5 architecture
def create_lenet5():
    model = keras.Sequential([
        # C1: Convolutional layer - 6 feature maps, 5x5 kernel
        layers.Conv2D(6, kernel_size=5, activation='tanh', input_shape=(32, 32, 1)),
        
        # S2: Subsampling layer - 2x2 average pooling
        layers.AveragePooling2D(pool_size=2, strides=2),
        
        # C3: Convolutional layer - 16 feature maps, 5x5 kernel
        layers.Conv2D(16, kernel_size=5, activation='tanh'),
        
        # S4: Subsampling layer - 2x2 average pooling
        layers.AveragePooling2D(pool_size=2, strides=2),
        
        # C5: Convolutional layer - 120 feature maps, 5x5 kernel
        layers.Conv2D(120, kernel_size=5, activation='tanh'),
        
        # Flatten for fully connected layers
        layers.Flatten(),
        
        # F6: Fully connected layer - 84 neurons
        layers.Dense(84, activation='tanh'),
        
        # Output layer - 10 neurons for digit classification
        layers.Dense(10, activation='softmax')
    ])
    
    return model

# Create and compile the model
model = create_lenet5()
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Display model architecture
model.summary()
```

```python
# Train the model and measure training time
start_time = time.time()

history = model.fit(x_train, y_train,
                    batch_size=128,
                    epochs=10,
                    validation_data=(x_test, y_test),
                    verbose=1)

training_time = time.time() - start_time

# Evaluate the model
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)

print(f"\n=== LeNet-5 Results ===")
print(f"Training Time: {training_time:.2f} seconds")
print(f"Test Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print(f"Test Loss: {test_loss:.4f}")
```

```python
# Plot training history
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

## 7. Use a pre-trained VGG16 model (via transfer learning) on a small custom dataset (e.g., flowers or animals). Replace the top layers and fine-tune the model. Include your code and result discussion.

```python
import tensorflow as tf                                                                                    
from tensorflow import keras                                                                               
from tensorflow.keras import layers                                                                        
from tensorflow.keras.applications import VGG16                                                            
import numpy as np                                                                                         
import matplotlib.pyplot as plt                                                                            
                                                                                                           
# Load and prepare a smaller dataset                                                                       
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()                                  
                                                                                                           
# Use only 3 classes for faster training (cats, dogs, birds)                                               
selected_classes = [2, 3, 5]  # bird, cat, dog                                                             
class_names = ['bird', 'cat', 'dog']                                                                       
                                                                                                           
# Filter data                                                                                              
train_mask = np.isin(y_train.flatten(), selected_classes)                                                  
test_mask = np.isin(y_test.flatten(), selected_classes)                                                    
                                                                                                           
x_train_subset = x_train[train_mask][:3000]  # Only 3000 samples                                           
y_train_subset = y_train[train_mask][:3000]                                                                
x_test_subset = x_test[test_mask][:600]      # Only 600 test samples                                       
y_test_subset = y_test[test_mask][:600]                                                                    
                                                                                                           
# Remap labels                                                                                             
label_mapping = {selected_classes[i]: i for i in range(len(selected_classes))}                             
y_train_subset = np.array([label_mapping[label[0]] for label in y_train_subset])                           
y_test_subset = np.array([label_mapping[label[0]] for label in y_test_subset])                             
                                                                                                           
# Resize to smaller images                                                                                 
x_train_subset = tf.image.resize(x_train_subset, [128, 128]) / 255.0                                       
x_test_subset = tf.image.resize(x_test_subset, [128, 128]) / 255.0                                         
                                                                                                           
# Convert to categorical                                                                                   
y_train_subset = keras.utils.to_categorical(y_train_subset, 3)                                             
y_test_subset = keras.utils.to_categorical(y_test_subset, 3)                                               
                                                                                                           
print(f"Training data shape: {x_train_subset.shape}")                                                      
print(f"Test data shape: {x_test_subset.shape}")                                                           
                                                                                                           
# Create simplified model                                                                                  
base_model = VGG16(weights='imagenet',                                                                     
                   include_top=False,                                                                      
                   input_shape=(128, 128, 3))                                                              
base_model.trainable = False                                                                               
                                                                                                           
model = keras.Sequential([                                                                                 
    base_model,                                                                                            
    layers.GlobalAveragePooling2D(),                                                                       
    layers.Dense(64, activation='relu'),                                                                   
    layers.Dropout(0.5),                                                                                   
    layers.Dense(3, activation='softmax')                                                                  
])                                                                                                         
                                                                                                           
model.compile(optimizer='adam',                                                                            
              loss='categorical_crossentropy',                                                             
              metrics=['accuracy'])                                                                        
                                                                                                           
# Fast training                                                                                            
print("Training with frozen base layers (fast version)...")                                                
history1 = model.fit(x_train_subset, y_train_subset,                                                       
                     batch_size=64,                                                                        
                     epochs=3,                                                                             
                     validation_data=(x_test_subset, y_test_subset),                                       
                     verbose=1)                                                                            
                                                                                                           
# Evaluate                                                                                                 
test_loss1, test_accuracy1 = model.evaluate(x_test_subset, y_test_subset, verbose=0)                       
print(f"Accuracy after initial training: {test_accuracy1:.4f} ({test_accuracy1*100:.2f}%)")                
                                                                                                           
# Fine-tuning (optional, also fast)                                                                        
base_model.trainable = True                                                                                
for layer in base_model.layers[:-2]:                                                                       
    layer.trainable = False                                                                                
                                                                                                           
model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.0001),                                       
              loss='categorical_crossentropy',                                                             
              metrics=['accuracy'])                                                                        
                                                                                                           
print("Fine-tuning (fast version)...")                                                                     
history2 = model.fit(x_train_subset, y_train_subset,                                                       
                     batch_size=64,                                                                        
                     epochs=2,                                                                             
                     validation_data=(x_test_subset, y_test_subset),                                       
                     verbose=1)                                                                            
                                                                                                           
test_loss2, test_accuracy2 = model.evaluate(x_test_subset, y_test_subset, verbose=0)                       
print(f"\n=== Transfer Learning Results (Fast Version) ===")                                               
print(f"Accuracy after initial training: {test_accuracy1:.4f} ({test_accuracy1*100:.2f}%)")                
print(f"Accuracy after fine-tuning: {test_accuracy2:.4f} ({test_accuracy2*100:.2f}%)")                     
print(f"Improvement: {(test_accuracy2-test_accuracy1)*100:.2f}%")                                          
                                                                                                           
# Plot results                                                                                             
plt.figure(figsize=(12, 4))                                                                                
                                                                                                           
acc = history1.history['accuracy'] + history2.history['accuracy']                                          
val_acc = history1.history['val_accuracy'] + history2.history['val_accuracy']                              
epochs = range(1, len(acc) + 1)                                                                            
                                                                                                           
plt.subplot(1, 2, 1)                                                                                       
plt.plot(epochs, acc, 'b-', label='Training Accuracy')                                                     
plt.plot(epochs, val_acc, 'r-', label='Validation Accuracy')                                               
plt.axvline(x=3, color='g', linestyle='--', label='Fine-tuning starts')                                    
plt.title('Transfer Learning - Model Accuracy')                                                            
plt.xlabel('Epoch')                                                                                        
plt.ylabel('Accuracy')                                                                                     
plt.legend()                                                                                               
                                                                                                           
plt.subplot(1, 2, 2)                                                                                       
# Show some predictions                                                                                    
predictions = model.predict(x_test_subset[:6])                                                             
predicted_classes = np.argmax(predictions, axis=1)                                                         
true_classes = np.argmax(y_test_subset[:6], axis=1)                                                        
                                                                                                           
for i in range(6):                                                                                         
    plt.subplot(2, 3, i+1)                                                                                 
    plt.imshow(x_test_subset[i])                                                                           
    plt.title(f'True: {class_names[true_classes[i]]}\nPred: {class_names[predicted_classes[i]]}')          
    plt.axis('off')                                                                                        
                                                                                                           
plt.tight_layout()                                                                                         
plt.show()     
```

- Initial training with frozen layers allows the model to adapt the pre-trained features to our specific task
- Fine-tuning further improves performance by allowing the model to adjust high-level features
- The pre-trained ImageNet features provide excellent starting point for animal classification
- Using transfer learning reduces training time and achieves better accuracy with limited data

## 9. Train a GoogLeNet (Inception v1) or its variant using a standard dataset like CIFAR-10. Plot the training and validation accuracy over epochs and analyze overfitting or underfitting.

```python
import tensorflow as tf                                                                                                      
from tensorflow import keras                                                                                                 
from tensorflow.keras import layers                                                                                          
import matplotlib.pyplot as plt                                                                                              
import numpy as np                                                                                                           
                                                                                                                             
# Load CIFAR-10 dataset                                                                                                      
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()                                                    
                                                                                                                             
# Use subset for faster training                                                                                             
x_train = x_train[:10000] / 255.0                                                                                            
y_train = y_train[:10000]                                                                                                    
x_test = x_test[:2000] / 255.0                                                                                               
y_test = y_test[:2000]                                                                                                       
                                                                                                                             
# Convert to categorical                                                                                                     
y_train = keras.utils.to_categorical(y_train, 10)                                                                            
y_test = keras.utils.to_categorical(y_test, 10)                                                                              
                                                                                                                             
print(f"Training data shape: {x_train.shape}")                                                                               
print(f"Test data shape: {x_test.shape}")                                                                                    
                                                                                                                             
# Simple Inception-like block                                                                                                
def inception_block(x, filters):                                                                                             
    # 1x1 conv                                                                                                               
    conv1 = layers.Conv2D(filters[0], (1, 1), activation='relu', padding='same')(x)                                          
                                                                                                                             
    # 3x3 conv                                                                                                               
    conv3 = layers.Conv2D(filters[1], (1, 1), activation='relu', padding='same')(x)                                          
    conv3 = layers.Conv2D(filters[2], (3, 3), activation='relu', padding='same')(conv3)                                      
                                                                                                                             
    # 5x5 conv (using two 3x3 for efficiency)                                                                                
    conv5 = layers.Conv2D(filters[3], (1, 1), activation='relu', padding='same')(x)                                          
    conv5 = layers.Conv2D(filters[4], (3, 3), activation='relu', padding='same')(conv5)                                      
    conv5 = layers.Conv2D(filters[5], (3, 3), activation='relu', padding='same')(conv5)                                      
                                                                                                                             
    # Max pooling                                                                                                            
    pool = layers.MaxPooling2D((3, 3), strides=(1, 1), padding='same')(x)                                                    
    pool = layers.Conv2D(filters[6], (1, 1), activation='relu', padding='same')(pool)                                        
                                                                                                                             
    # Concatenate all branches                                                                                               
    output = layers.concatenate([conv1, conv3, conv5, pool], axis=-1)                                                        
    return output                                                                                                            
                                                                                                                             
# Build simplified Inception model                                                                                           
def create_inception_model():                                                                                                
    inputs = layers.Input(shape=(32, 32, 3))                                                                                 
                                                                                                                             
    # Initial convolution                                                                                                    
    x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(inputs)                                                 
    x = layers.MaxPooling2D((2, 2))(x)                                                                                       
                                                                                                                             
    # Inception blocks (simplified)                                                                                          
    x = inception_block(x, [16, 16, 32, 8, 16, 16, 8])                                                                       
    x = layers.MaxPooling2D((2, 2))(x)                                                                                       
                                                                                                                             
    x = inception_block(x, [32, 32, 64, 16, 32, 32, 16])                                                                     
    x = layers.MaxPooling2D((2, 2))(x)                                                                                       
                                                                                                                             
    # Global average pooling                                                                                                 
    x = layers.GlobalAveragePooling2D()(x)                                                                                   
    x = layers.Dropout(0.5)(x)                                                                                               
                                                                                                                             
    # Output layer                                                                                                           
    outputs = layers.Dense(10, activation='softmax')(x)                                                                      
                                                                                                                             
    return keras.Model(inputs, outputs)                                                                                      
                                                                                                                             
# Create and compile model                                                                                                   
model = create_inception_model()                                                                                             
model.compile(optimizer='adam',                                                                                              
              loss='categorical_crossentropy',                                                                               
              metrics=['accuracy'])                                                                                          
                                                                                                                             
print("Model Summary:")                                                                                                      
model.summary()                                                                                                              
                                                                                                                             
# Train model (fast version)                                                                                                 
print("\nTraining Inception-like model...")                                                                                  
history = model.fit(x_train, y_train,                                                                                        
                    batch_size=128,                                                                                          
                    epochs=10,                                                                                               
                    validation_data=(x_test, y_test),                                                                        
                    verbose=1)                                                                                               
                                                                                                                             
# Evaluate model                                                                                                             
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)                                                         
                                                                                                                             
print(f"\n=== Inception Model Results ===")                                                                                  
print(f"Test Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")                                                      
print(f"Test Loss: {test_loss:.4f}")                                                                                         
                                                                                                                             
# Plot training history                                                                                                      
plt.figure(figsize=(12, 4))                                                                                                  
                                                                                                                             
plt.subplot(1, 2, 1)                                                                                                         
plt.plot(history.history['accuracy'], label='Training Accuracy')                                                             
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')                                                       
plt.title('Model Accuracy')                                                                                                  
plt.xlabel('Epoch')                                                                                                          
plt.ylabel('Accuracy')                                                                                                       
plt.legend()                                                                                                                 
                                                                                                                             
plt.subplot(1, 2, 2)                                                                                                         
plt.plot(history.history['loss'], label='Training Loss')                                                                     
plt.plot(history.history['val_loss'], label='Validation Loss')                                                               
plt.title('Model Loss')                                                                                                      
plt.xlabel('Epoch')                                                                                                          
plt.ylabel('Loss')                                                                                                           
plt.legend()                                                                                                                 
                                                                                                                             
plt.tight_layout()                                                                                                           
plt.show()                                                                                                                   
                                                                                                                             
# Analyze overfitting/underfitting                                                                                           
final_train_acc = history.history['accuracy'][-1]                                                                            
final_val_acc = history.history['val_accuracy'][-1]                                                                          
gap = final_train_acc - final_val_acc                                                                                        
                                                                                                                             
print(f"\n=== Overfitting/Underfitting Analysis ===")                                                                        
print(f"Final Training Accuracy: {final_train_acc:.4f}")                                                                     
print(f"Final Validation Accuracy: {final_val_acc:.4f}")                                                                     
print(f"Accuracy Gap: {gap:.4f}")                                                                                            
                                                                                                                             
if gap > 0.1:                                                                                                                
    print("Analysis: Model shows signs of OVERFITTING")                                                                      
    print("- Training accuracy significantly higher than validation")                                                        
    print("- Recommendations: More regularization, dropout, early stopping")                                                 
elif final_val_acc < 0.6:                                                                                                    
    print("Analysis: Model shows signs of UNDERFITTING")                                                                     
    print("- Both accuracies are low")                                                                                       
    print("- Recommendations: Larger model, more training, better features")                                                 
else:                                                                                                                        
    print("Analysis: Model shows good balance")                                                                              
    print("- Training and validation accuracies are close")                                                                  
    print("- Model generalizes well to unseen data")                                                                         
                                                                                                                             
# Additional analysis                                                                                                        
if history.history['val_loss'][-1] > history.history['val_loss'][5]:                                                         
    print("- Validation loss increasing: Early stopping recommended")                                                        
else:                                                                                                                        
    print("- Validation loss stable: Training progressing well")                                                             
                                                                                                                             
```

  - Less prone to overfitting compared to VGG due to better regularization

  - ResNet vs VGG: ResNet has fewer parameters and better gradient flow, reducing overfitting risk

  - ResNet vs AlexNet: ResNet is much deeper and captures more complex features

  - Start with ResNet pre-trained on ImageNet
  - Replace final classification layer with 3-class output (Normal, Pneumonia, COVID-19)
